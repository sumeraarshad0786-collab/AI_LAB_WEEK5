## Task 1 — Mini Intent Classifier (Keyword Rules)

**Goal:** Classify a user message into simple intents like: `greeting`, `goodbye`, `thanks`, `food`, `study`, `unknown`.

This cell defines `classify_intent(text)` and demonstrates it with sample messages (non-interactive demo).

```python
intent_keywords = {
    'greeting': ['hi', 'hello', 'salam', 'assalam', 'hey'],
    'goodbye':  ['bye', 'goodbye', 'see you', 'take care'],
    'thanks':   ['thanks', 'thank you', 'shukriya'],
    'food':     ['eat', 'restaurant', 'food', 'lunch', 'dinner'],
    'study':    ['study', 'exam', 'assignment', 'quiz', 'lecture']
}

def classify_intent(text):
    t = text.lower().strip()
    for intent in ['greeting', 'goodbye', 'thanks', 'food', 'study']:
        for kw in intent_keywords[intent]:
            if kw in t:
                return intent
    return 'unknown'

# Demo (non-interactive)
sample_msgs = [
    'Hi there!', 
    'Can you help with my assignment?', 
    "I'm hungry, where to eat?", 
    'Thanks a lot!', 
    'Bye, see you.' , 
    "What's this?"
]
print('Demo: classify_intent outputs:')
for m in sample_msgs:
    print(f"  Input: {m!r:30} -> Intent: {classify_intent(m)}")
```

## Task 2 — Tiny Expert System: Scholarship Eligibility

**Goal:** Decide scholarship status based on CGPA, family income, and achievements. Includes helper input readers and a demo.


```python
def decide_scholarship(cgpa, family_income, achievements):
    ach = [a.strip().lower() for a in achievements]
    has_national = any('national' in a for a in ach)
    
    if cgpa >= 3.7 and family_income <= 60000:
        return 'Full Scholarship'
    elif cgpa >= 3.0 and family_income <= 120000:
        return 'Partial Scholarship'
    elif cgpa >= 3.5 and has_national:
        return 'Partial Scholarship'
    else:
        return 'Not Eligible'

# Demo cases
demo_cases = [
    (3.8, 50000, ['sports national']),
    (3.2, 80000, []),
    (3.6, 200000, ['national olympiad']),
    (2.9, 40000, [])
]
print('Scholarship decisions (demo):')
for cgpa, inc, ach in demo_cases:
    print(f"  CGPA={cgpa}, income={inc}, achievements={ach} -> {decide_scholarship(cgpa, inc, ach)}")
```

## Task 3 — Rule-Based Recommender (Weighted Scoring)

**Goal:** Recommend an activity ('read', 'walk', 'nap', 'music') based on how the user feels and the time of day.


```python
def activity_recommender(feeling, time_of_day):
    feeling = feeling.lower().strip()
    time_of_day = time_of_day.lower().strip()
    scores = { 'read': 0, 'walk': 0, 'nap': 0, 'music': 0 }
    
    # Based on feeling
    if feeling == 'tired' or 'tired' in feeling:
        scores['nap'] += 3; scores['music'] += 1
    elif feeling == 'stressed' or 'stressed' in feeling:
        scores['walk'] += 2; scores['music'] += 2
    elif feeling == 'bored' or 'bored' in feeling:
        scores['read'] += 2; scores['walk'] += 1
    elif feeling == 'energetic' or 'energetic' in feeling:
        scores['walk'] += 3; scores['read'] += 1
    
    # Based on time
    if time_of_day in ('morning', 'afternoon'):
        scores['walk'] += 1; scores['read'] += 1
    elif time_of_day in ('evening',):
        scores['music'] += 2; scores['walk'] += 1
    elif time_of_day in ('night',):
        scores['nap'] += 2; scores['read'] += 1
    
    best = max(scores, key=scores.get)
    return best, scores

# Demo runs
demos = [
    ('tired', 'night'),
    ('stressed', 'evening'),
    ('bored', 'morning'),
    ('energetic', 'afternoon')
]
print('Activity recommender demo:') 
for f, t in demos:
    rec, sc = activity_recommender(f, t)
    print(f"  Feeling={f:9s}, Time={t:8s} -> Recommend: {rec:5s} | Scores: {sc}")
```

## Task 4 — Mini Finite‑State Machine (FSM): Traffic Light

**Goal:** Simulate a traffic light cycling through states with short durations. A demo run (no real delays) is shown below.

```python
import time
fsm = {
    'RED':    {'next': 'GREEN',  'duration': 1},
    'GREEN':  {'next': 'YELLOW', 'duration': 1},
    'YELLOW': {'next': 'RED',    'duration': 1},
}
def run_traffic_light(cycles=2, real_delay=False):
    state = 'RED'
    history = []
    for _ in range(cycles * len(fsm)):
        info = fsm[state]
        # For demo we don't wait if real_delay=False
        history.append(state)
        if real_delay:
            time.sleep(info['duration'])
        state = info['next']
    return history

# Demo
print('Traffic light demo (2 cycles):', run_traffic_light(2, real_delay=False))
```

## Task 5 — Robust Input Helpers (Re-usable Functions)

Helper functions: `read_nonempty`, `read_choice`, `read_float_range`. Demo shows usage with sample inputs (non-interactive).

```python
def read_nonempty_sim(s):
    s = s.strip()
    if s:
        return s
    raise ValueError('Empty input')

def read_choice_sim(s, choices):
    s_lower = s.strip().lower()
    choices_lower = [c.lower() for c in choices]
    if s_lower in choices_lower:
        return s_lower
    raise ValueError('Invalid choice')

def read_float_range_sim(s, lo, hi):
    x = float(s)
    if lo <= x <= hi:
        return x
    raise ValueError('Out of range')

# Demo (simulated inputs)
name = read_nonempty_sim('Fiza')
activity = read_choice_sim('read', ['read','walk','nap','music'])
cgpa = read_float_range_sim('3.45', 0.0, 4.0)
print(f"Demo form -> Name: {name}, Activity: {activity}, CGPA: {cgpa}")
```

## Wrap‑Up Challenge — AI Help Desk (Everything Together)

This small bot uses the functions above to classify intents and reply. A demo conversation is simulated below. Conversation history is printed at the end.

```python
history = []

def bot_reply(user_msg):
    intent = classify_intent(user_msg)
    if intent == 'greeting':
        reply = 'Hello! How can I help you today?'
    elif intent == 'thanks':
        reply = "You're welcome! Happy to help."
    elif intent == 'goodbye':
        reply = 'Goodbye! See you soon.'
    elif intent == 'study':
        rec, _ = activity_recommender('stressed', 'evening')
        reply = f'Study tip: set a 25-minute focus timer. Also try a quick {rec}.'
    elif intent == 'food':
        reply = 'Try a light, healthy meal so your energy stays stable while studying.'
    else:
        reply = "I'm not sure yet, but I can help with study tips, greetings, food suggestions, and simple activities."
    return intent, reply

# Simulated conversation (demo)
sim_msgs = ['Hi Fiza', "I'm bored", 'Can you give study tips?', 'Thanks!', 'Bye']

print('Simulated chat:') 
for m in sim_msgs:
    it, r = bot_reply(m)
    history.append((m, it, r))
    print(f"  You: {m}") 
    print(f"  Bot: {r}\n")

print('Conversation Summary:') 
for i, (u, it, r) in enumerate(history, 1):
    print(f"{i}. Intent={it:8s} | User='{u}' | Reply='{r}'")
```
